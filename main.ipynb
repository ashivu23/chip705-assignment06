```python
# Descriptive statistics for patients with eGFR <= 65

patients_needing_screening = pd.read_csv(
    "patients_needing_screening.csv"
)

stats = patients_needing_screening[
    ["Patient Age", "Patient Height", "Patient Weight"]
].describe()

stats
```

```python
# Average eGFR by BMI category

bmi_egfr = patients_needing_screening.groupby(
    "BMI Category"
)["Patient eGFR"].mean()

bmi_egfr.plot(kind="bar")

plt.title("Average eGFR by BMI Category")
plt.xlabel("BMI Category")
plt.ylabel("Average eGFR")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scatter plot: Patient Age vs. eGFR

plt.scatter(
    patients_needing_screening["Patient Age"],
    patients_needing_screening["Patient eGFR"]
)

plt.title("Patient Age vs. eGFR")
plt.xlabel("Patient Age (years)")
plt.ylabel("eGFR")
plt.tight_layout()
plt.show()
```

```python
# Pie chart: Patient Sex Distribution

sex_counts = patients_needing_screening["Patient Sex"].value_counts()

sex_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Patient Sex Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()
```
